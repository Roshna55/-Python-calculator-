```python
num1 = float(input("Enter the first number: "))
operator = str(input("Enter the operator you would like to use (+, -, *, /): "))
num2 = float(input("Enter the second number: "))

if operator == "+":
  print(num1 + num2)
elif operator == "-":
  print(num1-num2)

elif operator == "*":
  print(num1*num2)

elif operator == "/":
  if num2 == 0:
    print("Error")
  else:
     print(num1/num2)

else:
  print("Error")
```
